## Task 2: Basic CodeLlama Interaction

```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="codellama:latest", temperature=0)
```

```python
prompts = [
    "Write a Python function to check whether a number is prime.",
    "Explain this code"
]

for prompt in prompts:
    response = model.invoke(prompt)

    print("Prompt:", prompt)
    print("Response:", response.content)
    print()
```

## Task 3: Code Assistant Features

```python
def code_assistant(task, user_input):
    # Prompts with all features such as Generation, Explaination, Debugging and OPtimization of code
    prompts = {
        "Generate Code": f"Generate Python code for this requirement:\n{user_input}",
        "Explain Code": f"Explain this code step by step in simple language:\n{user_input}",
        "Debug Code": f"Find and fix bugs in this code:\n{user_input}",
        "Optimize Code": f"Improve the readability and efficiency of this code:\n{user_input}"
    }

    # If given task is out not in out prompts
    if task not in prompts:
        return "Invalid task selected."

    # Getting and returning the response
    response = model.invoke(prompts[task])
    return response.content
```

```python
result = code_assistant(
    "Generate Code",
    "Write a Python function to check whether a number is prime."
)

print(result)
```

```python
result = code_assistant(
    "Debug Code",
    """
def divide(a, b):
    return a / b

print(divide(10, 0))
"""
)

print(result)
```

## Task 5: Prompt Engineering for CodeLlama

```python
from langchain_core.prompts import PromptTemplate

# Create prompt templates
generate_code_prompt = PromptTemplate(
    input_variables=["requirement"],
    template="""
You are a Python coding assistant.
Write clean and readable Python code for the following requirement.
Include a brief explanation.

Requirement:
{requirement}
"""
)

explain_code_prompt = PromptTemplate(
    input_variables=["code"],
    template="""
You are a programming tutor.
Explain the following code step by step in simple language.

Code:
{code}
"""
)

debug_code_prompt = PromptTemplate(
    input_variables=["code"],
    template="""
You are a Python debugging assistant.
Identify the errors in the code, explain why they occur,
and provide the corrected code.

Code:
{code}
"""
)

optimize_code_prompt = PromptTemplate(
    input_variables=["code"],
    template="""
You are a Python code optimization assistant.
Improve the readability and efficiency of the code.
Provide the improved code and explain the changes.

Code:
{code}
"""
)

# Integrate templates into the Code Assistant
def code_assistant(task, user_input):
    templates = {
        "Generate Code": (generate_code_prompt, "requirement"),
        "Explain Code": (explain_code_prompt, "code"),
        "Debug Code": (debug_code_prompt, "code"),
        "Optimize Code": (optimize_code_prompt, "code")
    }

    if task not in templates:
        return "Invalid task selected."

    prompt_template, input_key = templates[task]
    prompt = prompt_template.format(**{input_key: user_input})

    response = model.invoke(prompt)
    return response.content
```

```python
# Test all features

tests = [
    ("Generate Code", "Write a function to calculate factorial."),
    ("Explain Code", "def square(n):\n    return n * n"),
    ("Debug Code", "def divide(a, b):\n    return a / b\n\nprint(divide(10, 0))"),
    ("Optimize Code", "result = []\nfor n in [1, 2, 3]:\n    result.append(n * 2)")
]

for task, user_input in tests:
    print(f"Task: {task}")
    print(code_assistant(task, user_input))
    print()
```
