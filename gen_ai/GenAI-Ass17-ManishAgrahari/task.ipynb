## Task 2 -> NLP Pipeline & Basic Text Cleaning -> Advanced Text Cleaning

```python
# Converting text to lowercase
df['content'] = df['content'].str.lower()
```

```python
import re

# Basic cleaning using re i.e. regular expression

def basic_clean(text):
    # Removing punctuation 
    text = re.sub(r'[!"#$%&\'()*+,-;=?[\\\]^_`{|}~]', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove extra whitespaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df['clean_text_basic'] = df['content'].apply(basic_clean)
```

## Task 3 -> Advanced Text Cleaning -> Removing Noise

```python
## Advanced Text Cleaning

def advance_cleaning(text):
    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    
    # Remove email
    text = re.sub(r'\S+@\S+', '', text)
    
    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)
    
    # Remove special characters and emojis
    text = re.sub(r'[^\w\s]', '', text)

    # Remove extra whitespaces created after above steps
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df["clean_text_advanced"] = df['clean_text_basic'].apply(advance_cleaning)
```

## Task 4 -> Handling Stopwords

```python
from nltk.corpus import stopwords

en_stop_words = stopwords.words('english')

def remove_stop_words(text):
    clean_words = []
    for word in text.split():
        if word not in en_stop_words:
            clean_words.append(word)
    return " ".join(clean_words)

df['text_no_stopwords'] = df['clean_text_advanced'].apply(remove_stop_words)
```

```python
df
```

## Task 5 -> Repeated Characters & Slang Normalization

```python
## For removing and normalizing repeated words
def normalize_repeated(text):
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    return text

## Slangs and its full expanded form in dictionary
slangs_dict = {
    "u": "you",
    "r": "are",
    "gr": "great",
    "idk": "i don't know",
    "asap": "as soon as possible",
    "omg": "oh my god",
    "bro": "brother"
}

## For expanding the slangs using slangs_dict
def expand_slangs(text):
    words = text.split()
    return " ".join([slangs_dict.get(word, word) for word in words])
```

```python
df['final'] = df['text_no_stopwords'].apply(normalize_repeated)
df['final'] = df['final'].apply(expand_slangs)
df
```

## Task 6 -> Tokenization

```python
from nltk.tokenize import word_tokenize

# Applying word_tokenize to final cleaned text and saving it to df with new column 
df['word_tokens'] = df['final'].apply(word_tokenize)
df[['final', 'word_tokens']].head(3)
```

## Task 7 -> Stemming

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

# function to apply stemming in our tokens which are in list
def apply_stemming(tokens):
    stemed_list = []
    for word in tokens:
        stemed_list.append(ps.stem(word))
    return " ".join(stemed_list)
```

```python
df["stemmed_tokens"] = df["word_tokens"].apply(apply_stemming)
df[["word_tokens", "stemmed_tokens"]].head(5)
```

## Task 8 -> Lemmatization

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

# function to apply lemmatization in our tokens which are in list
def apply_lemmatization(tokens):
    lemmatized_words = []
    for word in tokens:
        lemmatized_words.append(lemmatizer.lemmatize(word))
    return lemmatized_words
```

```python
df["lemmatized_tokens"] = df["word_tokens"].apply(apply_lemmatization)
df[["word_tokens", "lemmatized_tokens"]].head(5)
```

## Task 9 -> Final NLP pipeline creation

```python
## Starting with all import and initializing some packages

import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import  word_tokenize
from nltk.stem import WordNetLemmatizer

# Make sure resource are available if not downlaod using below commands
# nltk.download('punkt')
# nltk.download('punkt_tab')
# nltk.download('stopwords')
# nltk.download('wordnet')
# nltk.download('omw-1.4')

en_stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
slangs_dict = {
    "u": "you",
    "r": "are",
    "gr8": "great",
    "idk": "i don't know",
    "asap": "as soon as possible",
    "omg": "oh my god",
    "bro": "brother"
}
```

```python
def final_nlp_pipeline(text):

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove email addresses
    text = re.sub(r'\S+@\S+', '', text)
    
    # Remove HTML tags
    text = re.sub(r'<.*?>', '', text)
    
    # Remove numbers
    text = re.sub(r'\d+', '', text)
    
    # Removing punctuation 
    text = re.sub(r'[!"#$%&\'()*+,-;=?[\\\]^_`{|}~]', '', text)

    # Remove special characters and emojis
    text = re.sub(r'[^\w\s]', '', text)
    
    # Normalize repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    # Expand Slangs
    words = text.split()
    expanded_words = [slangs_dict.get(word, word) for word in words]

    # Stop word removal
    filtered_words = []
    for word in expanded_words:
        if word not in en_stop_words:
            filtered_words.append(word)

    # Tokenization
    tokens = word_tokenize(" ".join(filtered_words))

    # Lemmatization
    lemmatized_words = []
    for word in tokens:
        lemmatized_words.append(lemmatizer.lemmatize(word))

    return " ".join(lemmatized_words)
```

```python
df["final_processed_text"] = df["content"].apply(final_nlp_pipeline)
df[["content", "final_processed_text"]].head(10)
```

## Task 10 -> Observations & Insights

## 1.
#### Basic Cleaning handles common and simple text cleaning such as lowercase, punctuation, removing numbers,removing whitespaces etc.
#### Advanced Cleaning handles more specific types of noise such as URL's, emails, HTML tags, emojis, special characters, slangs etc.

## 2.
#### Lemmatization is preferred over stemming because it uses linguistic information to convert a word into its meaningful dictionary or its base form becuase
#### Stemming may produce a word that is not a valid English word because it mainly removes prefixes or suffixes using rules. Lemmatization generally produces a more meaningful base word.

## 3.
#### Preprocessing removes noise, normalizes text, and converts raw text into a consistent form. This helps subsequent NLP algorithms process text more effectively.
