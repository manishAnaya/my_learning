## Task 6 Why Text Splitting is Required

Large documents should not always be processed as a single piece of text because:

It may exceeds the context limit of embedding model

Large doc contains multiple topic and creating single representation for entire doc can make it difficult to get the relevant information.

## Problem Solved by Splitting

Splitting allows each meaningful section to be represented separately which leads to retrive meaninful insight which can be lost if splitting not done.

Processing smaller chunks is easy for computation.

Each chunk can be converted into its own numerical vector representation, allowing more precise similarity-based retrieval.

## Task 7 Character Text Splitter

```python
from langchain_text_splitters import CharacterTextSplitter

# Character-based text splitter
splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=20, separator='\n')

loaded_doc = PyPDFLoader('mensuration.pdf').load()

# Split the loaded document into smaller chunks
chunks = splitter.split_documents(loaded_doc)

print("Number of chunks:", len(chunks))

# Display all chunks
for i, chunk in enumerate(chunks):
    print(f"\nChunk {i + 1}")
    print(chunk.page_content)
```

## Task 8 Text Structure based Splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

rec_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=20)

loaded_doc = PyPDFLoader('mensuration.pdf').load()

# Split the loaded document into smaller chunks
rec_chunks = rec_splitter.split_documents(loaded_doc)

print("Number of chunks:", len(chunks))

for i, chunk in enumerate(rec_chunks):
    print(f"\nChunk {i + 1}")
    print(chunk.page_content)
```

CharacterTextSplitter generally uses a specified separator to split the document.

RecursiveCharacterTextSplitter uses multiple separators and tries them hierarchically

This allows RecursiveCharacterTextSplitter to preserve natural text structure better.

CharacterTextSplitter:
Document -> specified separator -> chunks

RecursiveCharacterTextSplitter:
Document -> paragraph -> line -> word -> character -> chunks

## Task 9 Document based structure splitting

```python
markdown_text = """
# Machine Learning

Machine Learning is a branch of Artificial Intelligence.

## Supervised Learning

Supervised learning uses labeled data to train a model.

### Classification

Classification is used when the target variable contains categories.
For example, spam detection is a classification problem.

### Regression

Regression is used when the target variable contains continuous values.
For example, house price prediction is a regression problem.

## Unsupervised Learning

Unsupervised learning works with unlabeled data.

### Clustering

Clustering groups similar data points together.
"""
```

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

headers_to_split_on = [
    ("#", "Header 1"),
    ("##", "Header 2"),
    ("###", "Header 3")
]
# MarkdownHeaderTextSplitter splits a Markdown document according to its heading hierarchy
splitter = MarkdownHeaderTextSplitter(headers_to_split_on)

# Splitting the markdown text
mark_chunks = splitter.split_text(markdown_text)

# Showing sample chunks
for i, chunk in enumerate(mark_chunks):
    print(f'chunk {i+1}')
    print(chunk.page_content)
    print(chunk.metadata)
    print()
```

## Task 10 Semantic Meaning based splitting

Semantic chunking splits a document based on the meaning or semantic similarity of the text rather than using only character length 
or document structure.

### How semantic chunking works

1. The document is divided into smaller units, such as sentences.
2. Each sentence is converted into an embedding (numerical vector).
3. The semantic similarity between neighboring sentences is calculated.
4. When there is a significant change in meaning, a new chunk is created.

### Role of Embeddings

Embeddings represent text as numerical vectors that capture semantic meaning.

Sentences with similar meanings tend to have similar vectors, while sentences discussing different topics tend to have vectors that are
farther apart.

Therefore, embeddings help identify where the meaning or topic of the document changes.

## Task 11 Unified Preprocessing Pipeline

```python
from langchain_community.document_loaders import DirectoryLoader, WebBaseLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

def load_and_split_documents(path_or_url):
    # If input is web url
    if path_or_url.startswith('http'):
        loader = WebBaseLoader(path_or_url)
    # otherwise treat it as file
    else :
        loader = DirectoryLoader(path_or_url, glob="*.pdf", loader_cls=PyPDFLoader)

    # Load the documents
    documents = loader.load()

    # Create a text splitter
    splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)

    # Split the documents
    chunks = splitter.split_documents(documents)

    return chunks
```

```python
## Task 11 test

url = "https://cloud.google.com/learn/what-is-artificial-intelligence"

web_chunks = load_and_split_documents(url)

print("Number of chunks:", len(web_chunks))

for i, chunk in enumerate(web_chunks[:5]):
    print(f"\nChunk {i + 1}")
    print(chunk.page_content)
```

## Task 12: Observations & Insights

### 1. Which loader is used for which data type?

TXT -> TextLoader
CSV -> CSVLoader
PDF -> PyPDFLoader
Directory -> DirectoryLoader
Webpage -> WebBaseLoader

### 2. Best splitter

Small text -> CharacterTextSplitter
Large PDFs -> RecursiveCharacterTextSplitter
Web data -> RecursiveCharacterTextSplitter
Structured Markdown -> MarkdownHeaderTextSplitter

### 3. Why is chunk overlap important?

Chunk overlap preserves context between consecutive chunks. It prevents important information from being lost when a sentence or concept crosses
a chunk boundary. This can improve the quality of retrieval in RAG systems.
