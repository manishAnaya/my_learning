## Task 3 Open AI vs HuggingFace Embedding models

1. When to prefer OpenAI embeddings?
OpenAI embeddings are preferred when we need easy integration, good
performance, scalability, and a managed cloud solution.

2. When to prefer Hugging Face embeddings?
Hugging Face embeddings are preferred when we need local execution,
privacy, lower cost, and more control over the embedding model.

3. Cost vs performance trade-offs
OpenAI requires API costs but provides easy setup and scalability.
Hugging Face can run locally without per-request API costs, but
requires local computational resources and model management.

# Task 4 Similarity Search APP

```python
from langchain_huggingface import HuggingFaceEmbeddings
from sklearn.metrics.pairwise import cosine_similarity

# converting document chunks using huggingface embedding as Open AI is not availble for free
embeddings = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')

doc_vectors = embeddings.embed_documents(documents)

def search(query, top=3):
    # converting query into embeddings
    query_vector = embeddings.embed_query(query)

    # Compare query vector with all document vectors
    scores = cosine_similarity([query_vector], doc_vectors)[0]

    best_indices = scores.argsort()[::-1][:top]
    
    return [(documents[i], scores[i]) for i in best_indices]
```

```python
## Task 4 Test

search("What is machine learning?", top=2)
```

```python
search('what is python', top=1)
```

```python
search('NLP helps in', top=3)
```

## Task 5 Similarity search with langchain abstraction

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore

# Using huggingface embedding
embeddings = HuggingFaceEmbeddings(model='sentence-transformers/all-MiniLM-L6-v2')

# Create LangChain vector store
vector_stores = InMemoryVectorStore(embeddings)

# Add documents
vector_stores.add_texts(documents)

query = 'what is NLP'

results = vector_stores.similarity_search(query, k=3)

print(results[0].page_content)
```

```python
from langchain_core.documents import Document

documents = [
    Document(page_content="Python is a popular, high-level programming language known for its readable syntax. It is used to build websites, automate repetitive tasks, analyze data, and create artificial intelligence applications."),
    Document(page_content="Machine learning is a branch of artificial intelligence that allows computers to learn patterns from data. After training on examples, a machine learning model can make predictions or decisions on new information."),
    Document(page_content="Deep learning is a type of machine learning that uses neural networks with many layers. These networks can learn complex patterns and are commonly used for image recognition, speech processing, and language tasks."),
    Document(page_content="Natural language processing, or NLP, helps computers understand and work with human language. It is used in applications such as chatbots, translation tools, sentiment analysis, and voice assistants."),
    Document(page_content="Data science combines statistics, programming, and domain knowledge to find useful insights in data. Data scientists clean and explore datasets, build models, and communicate their findings through reports and visualizations.")
]
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

# Taking OllamaEmbeddings and loading FAISS
embeddings = OllamaEmbeddings(model="nomic-embed-text")

db = FAISS.from_documents(documents, embedding=embeddings)

# Similarity search with query
results = db.similarity_search(query='What is data science', k=1)

print(results[0].page_content)
```

```python
# saving it to local
db.save_local("faiss_index")
```

```python
# reuse
db = FAISS.load_local("faiss_index", embeddings, allow_dangerous_deserialization=True)

results = db.similarity_search("What is machine learning?", k=3)

for doc in results:
    print(doc.page_content)
```

## Task 9 FAISS vs ChromaDB

FAISS and ChromaDB are both used for storing and retrieving vector embeddings.

FAISS is primarily a high-performance similarity-search library.
It is lightweight, fast, and suitable for local applications where the main requirement is efficient vector search. Persistence needs
to be handled explicitly using methods such as save_local() and load_local().

ChromaDB is a vector database designed for AI applications such as RAG. It provides persistent storage and supports documents,
embeddings, metadata, collections, and filtering.

FAISS is preferable when fast and simple local vector search is the main requirement. ChromaDB is preferable when building a persistent
RAG application that requires better document and metadata management.
