## Task 2: Basic OpenAI Q&A Chatbot

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq

load_dotenv()

# Open AI implementation if API key is there
# llm = ChatOpenAI(model="gpt-4o-mini")

# Initialize ChatGroq with model name
llm = ChatGroq(model='openai/gpt-oss-120b')

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    ("human", "{question}")
])

parser = StrOutputParser()

chain = prompt | llm | parser

questions = [
    "What is Python?",
    "What is machine learning?",
    "What is an API?",
    "What is an Gen AI?",
    "What is MCP?",
]

for question in questions:
    response = chain.invoke({'question': question})
    print(f"\nQuestion: {question}")
    print(f"Answer: {response}")
```

## Task 3: Multi-Turn Q&A

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

# Initialize ChatGroq with model name
llm = ChatGroq(model='openai/gpt-oss-120b')

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant. Use conversation history to answer follow-up questions."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{question}")
])

parser = StrOutputParser()

chain = prompt | llm | parser

history = []

while True:
    user_input = input('Ask Something...')
    if user_input.lower() == 'quit':
        break

    answer = chain.invoke({
        "history": history,
        "question": user_input
    })

    print("Assistant:", answer)

    history.append(HumanMessage(content=user_input))
    history.append(AIMessage(content=answer))
```

## Task 5: Ollama Chat Model with LangChain

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="gemma3:latest", temperature=0.5)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant. Response in short"),
    ("human", "{question}")
])

parser = StrOutputParser()

chain = prompt | llm | parser

questions = [
    "What is Python?",
    "What is machine learning?",
    "What is LangChain?"
]

for question in questions:
    print(f"\nQuestion: {question}")
    response = chain.invoke({'question': question})
    print(f"Answer: {response}")
```

## Task 7: Model Switch Logic



```python
prompt = ChatPromptTemplate([
    ("system", "You are a helpful AI assistant."),
    ("human", "{question}")
])

groq_llm = ChatGroq(model='openai/gpt-oss-120b')
ollama_llm = ChatOllama(model="gemma3:latest", temperature=0.5)

parser = StrOutputParser()

groq_chain = prompt | groq_llm | parser
ollama_chain = prompt | ollama_llm | parser

def get_answer(question: str, model_type='groq'):
    if model_type == 'groq':
        return groq_chain.invoke({"question": question})
    elif model_type == 'ollama':
        return ollama_chain.invoke({"question": question})
    else:
        raise ValueError("model_type must be 'groq' or 'ollama'")

question = input("Enter your question: ")
model_type = input("Choose model (groq/ollama): ").strip().lower()

try:
    print(get_answer(question, model_type))
except Exception as e:
    print("Error:", e)
```
