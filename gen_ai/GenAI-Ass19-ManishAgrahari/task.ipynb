## Task 1 -> Understanding Word Embeddings

#### Word Embeddings :
#### Word embedding is numerical representation of words in form of vectors.
#### Words that occur in similar context tends to have similar vector representation.
#### The vectors allow us to perform mathematical operations and calculate relationships between words.

#### for eg :
##### king  → [0.42, 0.81, -0.12, 0.65, ...]
##### queen → [0.39, 0.78, -0.10, 0.69, ...]
##### apple → [-0.21, 0.34, 0.87, -0.45, ...]

#### ------------------------------------------------------------------------------------------------------------------------------------
##### **One Hot Encoding and Bag of words both fails to capture sematic re;lationship between words which can be understand easily by example**

##### In One hot encoding suppose we have our vocabulary as:
##### ['cat', 'dog', 'king', 'queen']

##### one hot representation will be like
##### cat -> [1 0 0 0]
##### dog -> [0 1 0 0]
##### king -> [0 0 1 0]
##### queen -> [0 0 0 1]

#### from the above representation we can determine the semantic relationship between king and queen as [0 0 1 0] and [0 0 0 1], because every word is treated as independent.

##### **Bag of Words represents the document as word count**
#### BoW is useful for counting words, but it generally doesn't capture the meaning or relationship between words.
##### For Eg:
##### the dog chased the cat
##### the cat chased the dog

#### In above example we can see the order of words are diffrent thus have diffrent meaning, BoW ignores the word order and context. It only takes frequency of words

#### Word Embeddings solve these problems by learning vector representation based on context in which word appear.

#### For eg:
#### If our corpus contains
##### The dog eats food.
##### The dog plays in the park.
##### The dog runs quickly.

#### The model repeatedly sees words such as:
#### dog -> eat, dog -> food, dog -> plays, dog -> park, dog -> runs
#### the model learns a numerical representation for dog based on its surrounding words.

#### So we get something conceptually like this:
#### dog → [0.72, 0.41, -0.15, 0.63, ...]
#### cat → [0.70, 0.43, -0.12, 0.61, ...]

## Task 2 -> Word2Vec Overview

#### Word2Vec:
Word2Vec is a techniqueused to convert words into numerical vectors called word embeddings.

the main idea is that words which are used in similar context tends to have similar vector representation

#### Idea of predicting words from context
Word2Vec learns embeddings by using prediction tasks
In CBOW, the model predicts the target word from its surrounding context words, context word -> Target Word
In Skip-Gram, which is opposite of CBOW the model predicts surrounding context words from the target,  target word -> context word
word.

#### Vocabulary
Vocabulary is a collection of unique words learned from the training corpus

For example:

["the", "cat", "dog", "eats", "food"]

The vocabulary size is 5.

#### Context Window
Context Window determines how many words around the target word are considered during training

For example, with window = 2,

2 words before and 2 words after the target words will be considered as context

#### Embedding Dimension
Embedding Dimension is the number of numerical values use to represent each word

For example, vector_size = 100 means every word is represented by a
100-dimensional vector.

## Task 3 -> Types of Word2Vec Techniques

### Word2Vec mainly has two techniques:
#### CBOW — Continuous Bag of Words
#### Skip-Gram

### 1. CBOW: Continous bag of words
CBOW predicts the target word from surrounding context word

Example

Context => Target

The cat drinks milk

for target drinks context might be the, cat, milk

"the" + "cat" + "milk" → "drinks"

CBOW is useful when training speed is important and the corpus is large. 
It is generally efficient and works well for frequent words. 
It predicts a target word using its surrounding context words.

### 1. Skipgram
Skip-Gram predicts the surrounding context words using the target word.

Example

Target => Context

The cat drinks milk

"drinks" → "the" 

"drinks" → "cat"

"drinks" → "milk"

Skip-Gram is useful when we want better representations for less frequent or rare words. It predicts surrounding context words from a target word. It usually requires more training examples and can take longer to train than CBOW.

## Task 4 -> Neural Network Intuition Behind Word2Vec

1. Input Layer: 
Each word is initially represented using a one-hot encoded vector whose

size is equal to the vocabulary size.

Example:

cat → [1, 0, 0, 0]

2. Hidden Layer / Embedding Layer:
The hidden layer contains a smaller number of neurons called the embedding

dimension. During training, the model learns weights that capture

relationships between words.

3. Output Layer:
The output layer represents the vocabulary and predicts the target or 

context word depending on whether CBOW or Skip-Gram is being used.

4. Weights Become Word Embeddings:
The input-to-hidden layer contains a learned weight matrix. After training,

the row corresponding to a particular word is used as that word's

embedding vector.

For example:

cat → [0.21, 0.72, 0.15]

dog → [0.25, 0.68, 0.19]

Because cat and dog can occur in similar contexts, their learned vectors
can become similar.

Therefore:

Learned weights → Word vectors → Word embeddings

One-Hot Input --> Hidden Layer Weights --> Learned Weight Matrix --> Select the row for a word --> Word Embedding

## Task 5 -> Prepare Text for Word2Vec

```python
from nltk.tokenize import sent_tokenize, word_tokenize
import re

texts = [
    "This movie is really good. I enjoyed watching it.",
    "The story was amazing and the actors performed well.",
    "I really liked this movie. The acting was excellent.",
    "The movie has a beautiful story and great music."
]

sentences = []

for text in texts:
    sentences.extend(sent_tokenize(text))
print('Sentence:')
print(sentences)

tokenized_sentences = []

for sentence in sentences:
    # Lowercase
    sentence = sentence.lower()
    # Remove punctuation
    sentence = re.sub(r'[^\w\s]', '', sentence)
    # Word tokenization
    words = word_tokenize(sentence)
    tokenized_sentences.append(words)
print('\nTokenized Sentence:')
print(tokenized_sentences)

```

## Task 6 -> Train Word2Vec Model using CBOW

```python
from gensim.models import Word2Vec

# creating word2vec model with vector_size=100, window=5, min_count=1, sg=0 and tokenized_sentences from task 5
cbow_model = Word2Vec(vector_size=100, window=5, min_count=1, sg=0)

cbow_model.build_vocab(tokenized_sentences)

cbow_model.train(tokenized_sentences, total_examples=cbow_model.corpus_count, epochs=cbow_model.epochs)

# Vocabulary size
print("Vocabulary size:", len(cbow_model.wv))

# sample word
word = "movie"
embedding = cbow_model.wv[word]

print("Word:", word)
print("Embedding:", embedding)
print("Embedding Shape:", embedding.shape)
```

## Task 7 -> Train Word2Vec Model using Skip-Gram

```python
from gensim.models import Word2Vec

# creating word2vec model with vector_size=100, window=5, min_count=1, sg=1 and tokenized_sentences from task 5
skipgram_model = Word2Vec(vector_size=100, window=5, min_count=1, sg=1)

skipgram_model.build_vocab(tokenized_sentences)

skipgram_model.train(tokenized_sentences, total_examples=skipgram_model.corpus_count, epochs=skipgram_model.epochs)

# Vocabulary size
print("Vocabulary size:", len(skipgram_model.wv))

# sample word
word = "movie"
embedding = skipgram_model.wv[word]

print("Word:", word)
print("Embedding:", embedding)
print("Embedding Shape:", embedding.shape)
```

```python
cbow_similar = cbow_model.wv.most_similar("movie", topn=5)
skip_similar = skipgram_model.wv.most_similar("movie", topn=5)

print(cbow_similar)
print(skip_similar)

# For our small custom dataset, the measured training time for both CBOW and
# Skip-Gram was approximately 0 seconds. Therefore, the timing difference is
# not meaningful for this small dataset.
```

## Task 8 -> Word Similarity & Vector Operations

```python
#  Most similar words

cbow_similar = cbow_model.wv.most_similar("movie",topn=5)

skipgram_similar = skipgram_model.wv.most_similar("movie",topn=5)

print("CBOW Similar Words:")
print(cbow_similar)

print("\nSkip-Gram Similar Words:")
print(skipgram_similar)

#  Word arithmetic using words from our dataset
result = cbow_model.wv.most_similar(positive=["movie", "acting"],topn=5)
print('\nVector Arithmetic of movie + acting')
print(result)

movie_vector = cbow_model.wv["movie"]
acting_vector = cbow_model.wv["acting"]

if "movie" in cbow_model.wv and "acting" in cbow_model.wv:
    subtraction_vector = movie_vector - acting_vector
    print("\nClosest words to movie - acting:")
    print(
        cbow_model.wv.most_similar(
            positive=[subtraction_vector],
            topn=5
        )
    )

cbow_model.wv.similarity("movie", "story")

print(cbow_model.wv.index_to_key)

print(cbow_model.wv.doesnt_match(['movie', 'actors', 'music']))
```

## Task 10 -> OBSERVATIONS AND LIMITATIONS

## 1.
### CBOW (Continuous Bag of Words) 
It uses surrounding context words to predict the target word. It is generally faster and works well when words occur frequently.

### Skip-Gram 
It uses the target word to predict its surrounding context words. It usually takes more time to train but can perform better for rare words.

## 2.
#### Advantages of Word2Vec over TF-IDF
Word2Vec represents words as dense numerical vectors and can capture semantic relationships between words.

For example, words such as movie, film, actor, acting may have related vectors because they appear in similar contexts.

TF-IDF mainly tells us how important a word is in a document, but it does not directly capture the semantic relationship between words.

## 3.
#### Limitations of Word2Vec
Word2Vec requires a large amount of training data to learn good word representations.

It cannot properly represent unseen words that were not present in the training vocabulary.

Rare words may have less reliable embeddings, especially with a small dataset.

For example, the word "bank" can mean a financial institution or the side of a river, but Word2Vec generally uses one vector for both meanings.

## 4.
#### Context Still Matters in Modern NLP
The meaning of a word often depends on the surrounding words. 
For example:

I deposited money in the bank.

I sat near the river bank.

The word "bank" has different meanings in these two sentences.

Word2Vec generally gives the word bank one fixed representation, whereas modern Transformer models use the surrounding context to create context-dependent representations.

Therefore, understanding context remains a very important concept in modern NLP and is one of the key ideas behind Transformer-based models.
