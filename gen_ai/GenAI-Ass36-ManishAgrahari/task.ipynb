## HuggingFace Tasks

```python
import os
from dotenv import load_dotenv
from huggingface_hub import InferenceClient

load_dotenv()

token = os.getenv("HF_TOKEN")

# It initializes the connection to a specific model hosted in hugging face server
client = InferenceClient(token=token)

prompt = "Explain Generative AI in simple words."

response = client.chat_completion(
    model="google/flan-t5-base",
    messages=[
        {"role": "user", "content": prompt}
    ],
    max_tokens=150,
    temperature=0.7
)

print(response.choices[0].message.content)
```

```python

from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="""
    You are a helpful assistant.
    Explain {question} in simple words with one example.
    """,
    input_variables=["question"]
)

MODEL_NAME = "YOUR_SUPPORTED_CHAT_MODEL_ID"

questions = [
    "What is an AI Agent?",
    "What is Machine Learning?",
]

for question in questions:
    final_prompt = prompt_template.format(question=question)
    response = client.chat_completion(
        model=MODEL_NAME,
        messages=[
            {"role": "user", "content": final_prompt}
        ],
        max_tokens=150,
        temperature=0.7
    )

    print("Question:", question)
    print("Answer:", response.choices[0].message.content)
    print()
```

```python
from langchain_core.prompts import ChatPromptTemplate

# Create the chat prompt template

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Explain concepts in simple words with one example."),
    ("human", "{question}")
])

MODEL_NAME = "YOUR_SUPPORTED_CHAT_MODEL_ID"

questions = [
    "What is Generative AI?",
    "What is Retrieval-Augmented Generation (RAG)?"
]

# Test multiple questions
for question in questions:
    messages = chat_prompt_template.format_messages(question=question)

    formatted_messages = []

    for message in messages:
        if message.type == "system":
            role = "system"
        elif message.type == "human":
            role = "user"
        else:
            role = "assistant"

        formatted_messages.append({
            "role": role,
            "content": message.content
        })

    response = client.chat_completion(
        model=MODEL_NAME,
        messages=formatted_messages,
        max_tokens=150,
        temperature=0.7
    )

    print("Question:", question)
    print("Answer:", response.choices[0].message.content)
    print()
```

```python
from langchain_core.prompts import PromptTemplate

# Comparision of normal prompt and chat prompt

question = "What is Machine Learning?"

# Normal prompt template
normal_prompt = PromptTemplate(
    template="""
    You are a helpful assistant.
    Explain {question} in simple words with one example.
    """,
    input_variables=["question"]
)

final_prompt = normal_prompt.format(question=question)

normal_response = client.chat_completion(
    model=MODEL_NAME,
    messages=[
        {"role": "user", "content": final_prompt}
    ],
    max_tokens=150,
    temperature=0.7
)

# Chat prompt template
chat_messages = chat_prompt_template.format_messages(
    question=question
)

formatted_messages = [
    {
        "role": "system" if message.type == "system" else "user",
        "content": message.content
    }
    for message in chat_messages
]

chat_response = client.chat_completion(
    model=MODEL_NAME,
    messages=formatted_messages,
    max_tokens=150,
    temperature=0.7
)

print("Normal Prompt Response:")
print(normal_response.choices[0].message.content)

print("\nChat Prompt Response:")
print(chat_response.choices[0].message.content)
```
