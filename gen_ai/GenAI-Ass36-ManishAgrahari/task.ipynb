## HuggingFace Tasks

```python
import os
from dotenv import load_dotenv
from huggingface_hub import InferenceClient

load_dotenv()

token = os.getenv("HF_TOKEN")

# It initializes the connection to a specific model hosted in hugging face server
client = InferenceClient(token=token)

prompt = "Explain Generative AI in simple words."

response = client.chat_completion(
    model="mistralai/Mistral-7B-Instruct-v0.3",
    messages=[
        {"role": "user", "content": prompt}
    ],
    max_tokens=150,
    temperature=0.7,
    task="text-generation",
)

print(response.choices[0].message.content)
```

```python
import os
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_huggingface import HuggingFaceEndpoint
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

HF_TOKEN = os.getenv("HF_TOKEN")

# Create the model wrapper
llm = HuggingFaceEndpoint(
    repo_id="mistralai/Mistral-7B-Instruct-v0.3",
    task="text-generation",
    huggingfacehub_api_token=HF_TOKEN,
    max_new_tokens=150,
)

# Create the prompt template
prompt_template = PromptTemplate(
    template="""
    You are a helpful assistant.
    Explain {question} in simple words with one example.
    """,
    input_variables=["question"]
)

# Build the runnable chain
chain = prompt_template | llm | StrOutputParser()

questions = [
    "What is an AI Agent?",
    "What is Machine Learning?",
]

for question in questions:
    response = chain.invoke({"question": question})
    print("Question:", question)
    print("Answer:", response)
    print()
```

```python
from langchain_huggingface import ChatHuggingFace
from langchain_core.prompts import ChatPromptTemplate

# Create the chat model wrapper
chat_llm = ChatHuggingFace(llm=llm)

# Create the chat prompt template
chat_prompt = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful AI teacher. Explain concepts in simple words with one example.'),
    ('human', '{question}')
])

parser = StrOutputParser()

# Build the chat chain
chat_chain = chat_prompt | chat_llm | parser

# Test multiple questions
questions = [
    "What is a Large Language Model?",
    "What is Deep Learning?",
    "What is Retrieval-Augmented Generation?"
]

for question in questions:
    response = chat_chain.invoke({"question": question})

    print("Question:", question)
    print("Answer:", response)
    print("-" * 50)
```

```python
# Compare normal and chat prompts

question = "What is Machine Learning?"

normal_response = chain.invoke({
    "question": question
})

chat_response = chat_chain.invoke({
    "question": question
})

print("Normal Prompt Response:")
print(normal_response)

print("\nChat Prompt Response:")
print(chat_response)
```
