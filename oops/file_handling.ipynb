```python
def count_lines(file_name):
    with open(file_name, "r") as file:
        count = 0
        for line in file:
            count += 1
            print(line)
    print(count)

count_lines("new_file.txt")
        
```

```python
def count_words(file_name):
    with open(file_name, "r") as file:
        text = file.read()
        text_list = text.split()
        print(len(text_list))
```

```python
count_words("new_file.txt")
count_lines("new_file.txt")
```

```python
def count_char(file_name):
    with open(file_name, "r") as file:
        data = file.read()
        cleaned_data = data.replace(" ", '').replace("\n", "")
        print(len(data))
        print(len(cleaned_data))
```

```python
count_char("new_file.txt")
```
