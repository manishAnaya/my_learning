```python
pan = "ABCDE124F"

pattern = r"^[A-Z]{5}[0-9]{4}[A-Z]{1}$"

if re.match(pattern, pan):
    print("Valid PAN")
else:
    print("Invalid")
```

```python
mobile = "1876543210"

pattern = r"^[6-9]{1}[0-9]{9}$"

if re.match(pattern, mobile):
    print("Valid Number")
else:
    print("Invalid")
```

```python
text = "Manish lives in Bangalore and works at OpenAI"
pattern = r"\b[A-Z][a-z]*"
ans = re.findall(pattern, text)

print(ans)
```

```python
try:
    num1 = int(input("Enter 1st number: "))
    num2 = int(input("Enter 2nd number: "))
    div = num1 / num2
    print(div)
except ValueError:
    print("Please enter valid number")
except ZeroDivisionError:
    print("cannot divide by 0")
else:
    print("Answer is ", div)
finally:
    print("Program Ended Successfully...")
```
