```python
import re

text = "<h1>Hello my name is Manish</h1>"

re.sub(r'<.*?>', '', text)
```

```python
data = pd.DataFrame({"col": [
    'Hello <h1> Manish </h1>',
    '<a>click here</a>',
    '<p>How are you?</p>'
]})
```

```python
def removeTags(text):
    cleaned = []
    for i in text:
        cleaned.append(re.sub(r'<.*?>', '', i))
    return cleaned
```

```python
data['col'].apply(removeTags)
```

```python
slangs_dict = {
    "gm": "good morning",
    "fomo": "fear of missing out",
    "yolo": "you only live once",
    "asap": "as soon as possible"
}
slangs_dict
```

```python
slang_df = pd.DataFrame(
    {"col": ["Hey gm", "I have fomo of that", "Enjoy man because yolo", "Please come asap"]}
)
slang_df
```

```python
def expand_slangs(text):
    words = text.split()
    return " ".join([slangs_dict.get(word, word) for word in words])
```

```python
slang_df.col = slang_df.col.apply(expand_slangs)
slang_df
```

## Workign with Emojis cleaning

```python
import nltk

nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
len(stopwords.words('english'))
```

```python
english_stop_words = stopwords.words('english')
```

```python
msg = "Hii How are you I was looking for you.. I am quite exited to eat an apple"
```

```python
without_stopword = []
for i in msg.split():
    if i in english_stop_words:
        without_stopword.append('')
    else:
        without_stopword.append(i)
```

```python
" ".join(without_stopword)
```

```python
def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags
        "]+",
        flags=re.UNICODE
    )

    return emoji_pattern.sub(r'', text)
```

```python
data = {
    "content": [
        "Hello 😊, how are you?",
        "I am learning NLP 🚀 and Python 🐍",
        "This is amazing! 🔥🔥🔥",
        "Good morning ☀️ everyone!",
        "I love India 🇮🇳 ❤️"
    ]
}

data = pd.DataFrame(data)
data
```

```python
data.content.apply(remove_emojis)
```
