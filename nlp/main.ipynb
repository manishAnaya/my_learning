```python
import ast
```

```python
def extractGenres(col):
    genre = []
    for i in ast.literal_eval(col):
        genre.append(i['name'])
    return genre
```

```python
data.genres = data.genres.apply(extractGenres)
```

```python
data.keywords.tolist()[0]
```

```python
def extractKeywords(col):
    key = []
    for i in ast.literal_eval(col):
        key.append(i['name'])
    return key
```

```python
data.keywords = data.keywords.apply(extractKeywords)
```

```python
data.release_date = pd.to_datetime(data.release_date, errors='coerce')
```

```python
def extractCast(col):
    cast = []
    j = 0
    for i in ast.literal_eval(col):
        if j < 3:
            cast.append(i['name'])
            j += 1
    return cast
```

```python
data.cast = data.cast.apply(extractCast)
```

```python

ast.literal_eval(data.crew[0])
```

```python
def extractCrew(col):
    crew = []
    for i in ast.literal_eval(col):
        if i['job'] == 'Director':
            crew.append(i['name'])
            break
    return crew
```

```python
data.crew.apply(extractCrew)
```

```python
data.crew = data.crew.apply(extractCrew)
```

**Now we can proceed with Stop Word Removal and Stemming or Lemmetization..**
**After that Bag of Words**

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stemming(text):
    L = []
    for i in text.split():
        L.append(ps.stem(i))
    return " ".join(L)
```

```python
data.tags = data.tags.apply(stemming)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(data.tags)
vectors = vectors.toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out()[:200]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarities = cosine_similarity(vectors)
```

```python
similarities
```

```python
data[data.title == 'Avatar'].index[0]
```

```python
def recommend(movie):
    movie_index = data[data.title == movie].index[0]
    recommendation = similarities[movie_index]
    movie_list = sorted(enumerate(recommendation), reverse=True, key=lambda x: x[1])[1: 6]
    print(movie_list)
    for movies in movie_list:
        print(data.title[movies[0]])
```

```python
recommend('Avatar')
```

```python
sorted(enumerate(recommend('Avatar')), reverse=True, key=lambda x: x[1])[1: 6]
```
