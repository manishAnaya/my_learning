## TF-IDF
#### How important is this word to this document compared with the whole collection
#### TF-IDF = Term Frequency × Inverse Document Frequency

TF → Term Frequency
How frequently does a word appear inside one document?
TF = Number of times term appears
     ─────────────────────────────
     Total number of terms

IDF → Inverse Document Frequency
How rare/common is that word across all documents?
IDF = log(N / DF)
Where
N = total number of documents
DF = number of documents containing the term**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf = TfidfVectorizer()
```

```python
tf.fit_transform(df.col).toarray()
```

```python
tf.vocabulary_
```

```python
tf.get_feature_names_out()
```

## Going to train word2vec model on presonal data

```python
with open('J. K. Rowling - Harry Potter 1 - Sorcerer Stone.txt', 'rt') as b1:
    h1 = b1.read()

with open('J. K. Rowling - Harry Potter 2 - The Chamber Of Secrets.txt', 'rt') as b2:
    h2 = b2.read()

with open('J. K. Rowling - Harry Potter 3 - Prisoner of Azkaban.txt', 'rt') as b3:
    h3 = b3.read()

with open('J. K. Rowling - Harry Potter 4 - The Goblet of Fire.txt', 'rt') as b4:
    h4 = b4.read()
```

```python
data = h1 + h2 + h3 + h4
```

```python
len(data)
```

```python
len(data.split()) ## Total number of words in Harry Potter Data corpus
```

```python
## Applying NLP preprocessing 

## Tokenization -> Sentence Tokenization
## Stop word Removal 
## Lower casing

from gensim.utils import simple_preprocess
from nltk.tokenize import sent_tokenize
```

```python
potter = []
data = sent_tokenize(data)
```

```python
for i in data:
    potter.append(simple_preprocess(i))
```

```python
potter
```

```python
from nltk.corpus import stopwords

en_sw = set(stopwords.words('english'))
en_sw
```

```python
def remove_stopwords(text):
    cleaned_text = []
    for sentence in text:
        cleaned_sentence = []
        for word in sentence:
            if word not in en_sw:
                cleaned_sentence.append(word)
        cleaned_text.append(cleaned_sentence)
    return cleaned_text
```

```python
cleaned_potter = remove_stopwords(potter)
cleaned_potter
```

```python
import gensim

model = gensim.models.Word2Vec(window=10, min_count=2)
model
```

```python
model.build_vocab(cleaned_potter)
```

```python
model.train(cleaned_potter, total_examples=model.corpus_count, epochs=model.epochs)
```
