```python
from sklearn.feature_extraction.text import CountVectorizer
```

## N-Grams -> We take sequence of word in vocabulary i.e. It is a sequence of N consecutive words/tokens from a sentence.
Here we have bigram and trigram depends on the value of sequence like we take 2 sequence in bigram and 3 in trigram
sentence = w1 w2 w3 w4

bigram => w1w2 w2w3 w3w4
trigram => w1w2w3 w2w3w4
4-gram => w1w2w3w4

```python
cv = CountVectorizer(ngram_range=(1, 2))

cv.fit_transform(df.col).toarray()
```

```python
cv.vocabulary_
```

```python
## Bigram => 8
## unigram => 4
## trigram => 4
## unigram + Bigram => 12
```

## TF-IDF
#### How important is this word to this document compared with the whole collection
#### TF-IDF = Term Frequency × Inverse Document Frequency

TF → Term Frequency
How frequently does a word appear inside one document?
TF = Number of times term appears
     ─────────────────────────────
     Total number of terms

IDF → Inverse Document Frequency
How rare/common is that word across all documents?
IDF = log(N / DF)
Where
N = total number of documents
DF = number of documents containing the term**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf = TfidfVectorizer()
```

```python
tf.fit_transform(df.col).toarray()
```

```python
tf.vocabulary_
```

```python
tf.get_feature_names_out()
```

```python
tf.idf_
```
