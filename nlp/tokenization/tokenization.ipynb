## Now Step is Stemming and Lemmetization

#### Stemming simply cuts off prefixes/suffixes from a word to get its root form.
Example
Word	Stem
playing	play
played	play
studies	studi
studying	studi
happiness	happi
connection	connect

#### Lemmatization is more sophisticated.
Instead of simply cutting characters, it tries to find the actual dictionary/base form of a word, called its lemma.

Example
Word	Lemma
playing	play
played	play
studies	study
better	good
children	child
mice	mouse

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
words = [
    'Running', 'Walking', 'Stories', 'Runs', 'easily', 'fairly'
]
```

```python
for word in words:
    print(ps.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
from nltk.corpus import wordnet
```

```python
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
```

### pos is part of speech with v as verb, a = adverb and n as noun

```python
lemmatizer.lemmatize('running', pos='v')
```

```python
lemmatizer.lemmatize('better', pos='a')
```

```python
lemmatizer.lemmatize('mice', pos='n')
```
