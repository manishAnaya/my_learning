## Adding some extra columns:
#### df["Column Name"] = Scaler Value

```python
df['Age Group'] = df["Age"].apply(
    lambda x: "Young" if x < 21 else "Adult"
)
```

```python
df['Age Group'].value_counts()
```

```python
df["Performance"] = df["CGPA"].apply(
    lambda x: "Good" if x > 8.0 else "Average" 
)
```

```python
df['Performance'].value_counts()
```

```python
def performance(cgpa):
    if cgpa >= 8:
        return "Excellent"
    elif cgpa >= 6:
        return "Good"
    elif cgpa >= 4:
        return "Average"
    else:
        return "Needs Improvement"

df["Performance"] = df["CGPA"].apply(performance)
```

```python
df.info()
```

```python
df["ID"] = df["ID"].astype("int32")
```

```python
df["Age"] = df["Age"].astype("int32")
```

```python
columns = ["Academic Pressure", "Work Pressure", "CGPA", "Study Satisfaction", "Job Satisfaction", "Work/Study Hours", "Depression"]
df[columns] = df[columns].astype('int32')
```

```python
new_col = ["Gender", "City", "Profession", "Sleep Duration", "Dietary Habits", "Degree", "Have you ever had suicidal thoughts ?", "Family History of Mental Illness", "Age Group", "Performance"]
df[new_col].nunique()
```

```python
df[new_col] = df[new_col].astype("category")
```

```python
df['Financial Stress'].unique()
```

```python
df['Financial Stress'] = pd.to_numeric(
    df["Financial Stress"],
    errors="coerce"
)
```

```python
df['Financial Stress'] = df['Financial Stress'].astype('int32')
```
