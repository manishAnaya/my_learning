```python
import pandas as pd
import numpy as nm
```

```python
pd.Series([40, 50, 60], index=["Butter", "Milk", "Bread"], name="General Store Items")
```

```python
## Series from list

list = ["India", "China", "Russia", "USA"]
pd.Series(list)
```

```python
runs = [45, 78, 96, 114, 198, 48, 52]
pd.Series(runs, dtype="int32")
```

```python
subject = ["Maths", "Physics", "Chemistry", "Biology", "English"]
marks = [78, 65, 75, 84, 69]

pd.Series(data=marks, index=subject, dtype="int32", name="Pre Board Marks")
```

```python
## Series from numpy arrays
values = nm.arange(10, 21)
pd.Series(values)
```

```python
## Series from Dictionary
data = {
    "Maths": 78,
    "English": 87,
    "Science": 98,
    "Hindi": 72,
    "GK": 87,
}
pd.Series(data)
```

```python
## Series from Custom index
values = [1, 2, 3, 4, 5]
pd.Series(values, index=["a", "b", "c", "d", "e"])
```

```python
## Series from Scaler
pd.Series(4, index=["a", "b", "c", "d", "e"])
```

### Mathematical Fn in Pandas

#### 1. sum() ---> to get the sum of all values present in series.
#### 2. product() ---> to get the product of all value present in series.
#### 3. mean() ---> to get the mean i.e. average of all values.
#### 4. median() ---> to get the mean i.e. median of all values in series.
#### 5. mode() ---> to get the mean i.e. mode from all values in series i.e max time highest frequency.
#### 6. std() ---> to get thow much the series is distributed from the mean of values.
#### 7. min() ---> to get the minimum value from all value present in series.
#### 8. max() ---> to get the maxmum value from all value present in series.
#### 9. variance() ---> to get the insight about the distribution of data from mean value.. It is square od standard deviation
#### 10. describe() ---> to get the whole summary of important mathematical calculations.

# Indexing in Series:
#### In order to have the accesss over the certain values
##### 1. **Positive Indexing** -> start from 0 to forward direction upto the length of series - 1
##### 2. **Slicing** -> In order to get the certain values upto specific range i.e. slice of series -> [initial_val: final_val+1: interval]
##### 3. **Fancy Indexing** -> In order to get the specific value from the specific list of index
##### 3. **Indexing with label** -> In order to get the data directly describing the label name

```python
total_vid[::-1]
```

```python
total_vid[9]
```

```python
total_vid[1:9]
```

```python
total_vid[[4, 9, 15, 78]]
```
