```python
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import google.generativeai as genai
import os
```

```python
api_key = os.getenv('GEMINI_API_KEY')

genai.configure(api_key=api_key)
```

```python
# g_model = genai.GenerativeModel('gemini-3.5-flash-lite')
# g_model.generate_content('Hello')
```

```python
google_model = ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')

groq_model = ChatGroq(model='openai/gpt-oss-120b')
```

```python
prompt1 = PromptTemplate(template='Generate a detailed report on {Topic}', input_variables=['Topic'])

prompt2 = PromptTemplate(template='Create a five point summary on {Report}', input_variables=['Report'])
```

```python
final1 = prompt1.invoke({'Topic': 'Google vs Groq Models'})
```

```python
parser = StrOutputParser()
```

```python
chain = prompt1 | google_model | parser | prompt2 | groq_model | parser
```

```python
response = chain.invoke({'Topic': 'Google vs Groq Models'})
```
