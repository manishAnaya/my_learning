```python
nums = [0, 1, 0, 3 , 12]
output = []
count = 0

for num in nums:
    if num == 0:
        count += 1
        continue
    else:
        output.append(num)
for i in range(count):
    output.append(0)

print(output)
```

```python
numbers = [0, 1, 0, 3 , 12]

def moveZeroes(nums):
    m = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[m], nums[i] = nums[i], nums[m]
            m += 1
    return nums

moveZeroes(numbers)
```
