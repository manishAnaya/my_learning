```python
from langchain_core.runnables import RunnableSequence
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
from langchain_groq import ChatGroq
```

```python
load_dotenv()
```

```python
prompt = PromptTemplate(
    template='Generate a joke on {topic}',
    input_variables=['topic']
)
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b', temperature=1)
```

```python
parser = StrOutputParser()
```

```python
chain = RunnableSequence(prompt, llm, parser)
```

```python
result = chain.invoke({'topic': 'cricket'})
```

```python
print(result)
```

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough
```

```python
prompt1 = PromptTemplate(
    template='Generate a joke on {topic}',
    input_variables=['topic']
)
```

```python
chain_1 = RunnableSequence(prompt1, llm, parser)
```

```python
prompt2 = PromptTemplate(
    template='Provide explaination of the {joke}',
    input_variables=['joke']
)
```

```python
explain = prompt1 | llm | parser | prompt2 | llm | parser
```

```python
res = explain.invoke({'topic': 'AI vs Human'})
```

```python
print(res)
```

```python
parallel = RunnableParallel({'joke': RunnablePassthrough(), 'explaination': prompt2 | llm | parser})
```

```python
final_chain = chain | parallel
```

```python
res = final_chain.invoke({'topic': 'AI vs Human'})
```

```python
print(res['joke'])
```

```python
print(res['explaination'])
```
