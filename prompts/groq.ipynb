```python
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
model = ChatGroq(model='openai/gpt-oss-120b', temperature=0.8)
```

```python
parser = StrOutputParser()
```

```python
prompt = PromptTemplate(
    template='Translate the following text {text} from english to {mentioned} language',
    input_variables=['text', 'mentioned']
)
```

```python
chain = prompt | model | parser
```

```python
response = chain.invoke({'text': 'Hello How are you?', 'mentioned': 'French'})
```

```python
print(response)
```
