```python
from langchain_core.runnables import RunnablePassthrough, RunnableSequence, RunnableParallel, RunnableLambda
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser()
```

```python
prompt = PromptTemplate(
    template='Provide Explaination of topic {topic}',
    input_variables=['topic']
)
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b', temperature=0.8)
```

```python
chain_1 = RunnableSequence(prompt, llm, parser)
```

```python
def count_len(words):
    return len(words.split())
```

```python
chain_2 = RunnableParallel({'explaination': RunnablePassthrough(), 'word_count': RunnableLambda(count_len)})
```

```python
final_chain = RunnableSequence(chain_1, chain_2)
```

```python
response = final_chain.invoke({'topic': 'Water'})
```

```python
print(response)
```

```python
print(response['explaination'])
```

```python
print(response['word_count'])
```
