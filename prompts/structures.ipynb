```python
from pydantic import BaseModel, Field
from typing import Optional, Literal
from langchain_groq import ChatGroq
from dotenv import load_dotenv
```

```python
from typing import Optional
from pydantic import BaseModel, Field

class Review(BaseModel):
    Product: str
    Pros: list[str]
    Cons: list[str]
    Review: Optional[str] = None
    Ratings: float = Field(
        default=3,
        gt=0,
        lt=5,
        description="Rating given by the reviewer for the product",
    )
```

```python
load_dotenv()
```

```python
review = '''The Sony WH-1000XM5 headphones offer excellent noise cancellation, making them a great choice for travel, studying, or working in busy places. They are comfortable to wear for long periods, deliver clear sound, and have a battery that lasts through many hours of listening. On the downside, they are quite expensive, and their carrying case is larger and bulkier than some people might prefer.
'''
```

```python
class Data(BaseModel):
    Product: str = Field(description='Provide name of the product')
    Pros: list[str] = Field(description='Mention positive points of product')
    Cons: list[str] = Field(description='Mention negative points of product')
    Sentiment: Literal['Positive', 'Negative'] = Field(description='Provide the sentiment of user review')
    Ratings: float = Field(
        lt=5,
        gt=1,
        default=3,
        description='Rating given by the reviewer for the product'
    )
    Reviewer: Optional[str] = Field(description='Name of the reviewer', default=None)
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b', temperature=0.6)
```

```python
structured_llm = llm.with_structured_output(Data)
```

```python
res = structured_llm.invoke(review)
```
