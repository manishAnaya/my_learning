## Consusion Matrix is a matrix used to evaluate the performance of classification ML model by comparing models predicted value and actual values.
|                        |     **Actual Positive** |     **Actual Negative** |
| ---------------------- | ----------------------: | ----------------------: |
| **Predicted Positive** |  **TP** (True Positive) | **FP** (False Positive) |
| **Predicted Negative** | **FN** (False Negative) |  **TN** (True Negative) |

#### Accuracy -> out of all how many are correct 
#### Accuracy = TP + TN / TP + TN + FP + FN

#### Precision -> Of all predicted positive how many are actual positive
#### Precision = TP / TP + TN

#### Recall -> Of actual positives how many we correctly identified 
#### Recall = TP / TP + FN

#### Specificity -> Of Actual negative how many we correctly identified
#### Specificity = TN / TN + FP

#### F1 Score -> Balance between Precision and Recall
#### F1 Score = (2 x Precision x Recall) / Precision + Recall
