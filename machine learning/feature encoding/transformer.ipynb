```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, LabelEncoder, MinMaxScaler, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('Bank_Marketing_Original_Data.csv')
```

```python
df.head()
```

```python
df.info()
```

## In this we have 16 Columns

## To which i can divide into 4 major parts as:
#### Numerical cols -> Scaling using StandardScale or MinMaxScale -> ['Balance', 'Duration']
#### Binary Cols -> Encoding using LabelEncoder or just mapping (1, 0) -> ['Default', 'Housing', 'Loan', 'TARGET']
#### Ordinal Categorial Cols -> Encoding using OrdinalEncoder -> ['Education']
#### Nominal Categorical Cols -> Encoding using OneHotEncoder -> ['Job', 'Marital', 'Month', 'poutcome']

```python
oe = OrdinalEncoder(categories=[['unknown', 'primary', 'secondary', 'tertiary']])
df['Education'] = oe.fit_transform(df[['Education']])
```

```python
df
```

```python
num_cols = ['Balance', 'Day', 'Duration', 'Campaign', 'Pdays', 'Previous']

cat_cols = ['Job', 'Marital', 'Month', 'Housing', 'Loan', 'poutcome']
```

```python
transformer = ColumnTransformer(transformers=[('cat', OneHotEncoder(), cat_cols), ('num', StandardScaler(), num_cols)], remainder='passthrough')
```

```python
X = df.iloc[:, :-1]
```

```python
Y = df['TARGET']
```

```python
X = transformer.fit_transform(X)
```
