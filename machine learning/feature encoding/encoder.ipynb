## Ordinal Encoder => Education
## Label Encoder => Target
## One Hot Encoder (Nominal Categorical) => Married, .....

```python
ordinal_data = df[['Education']]
```

```python
label_data = df[['TARGET']]
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['unknown', 'primary', 'secondary', 'tertiary']])
```

```python
ordinal_data = oe.fit_transform(ordinal_data)
```

```python
ordinal_data
```

```python
df['Education'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
ld = le.fit_transform(label_data)
```

```python
ld
```
