## Auto Cleaning 

### Why Data is missing that have mainly 3 ways =>
#### Missing Completely at Random (MCAR)
#### Missing at Random (MAR)
#### Missing not at Random (MNAR)

```python
import pandas as pd
```

```python
df = pd.read_csv('unclean_smartwatch_health_data.csv')
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.drop(columns=['User ID'], inplace=True)
```

```python
df[df['Heart Rate (BPM)'].isnull() == True]
```

```python
df['Heart Rate (BPM)'].median()
```

```python
df['Heart Rate (BPM)'].describe()
```

```python
df['Heart Rate (BPM)'] = df['Heart Rate (BPM)'].fillna(df['Heart Rate (BPM)'].median())
```

```python
df[df['Blood Oxygen Level (%)'].isnull() == True]
```

```python
df['Blood Oxygen Level (%)'].describe()
```

```python
df['Blood Oxygen Level (%)'].median()
```

```python
df['Blood Oxygen Level (%)'] = df['Blood Oxygen Level (%)'].fillna(df['Blood Oxygen Level (%)'].mean())
```

```python
df['Step Count'].describe()
```

```python
df['Step Count'].median()
```

```python
df['Step Count'] = df['Step Count'].fillna(df['Step Count'].median())
```

```python
df['Sleep Duration (hours)'].describe()
```

```python
df['Sleep Duration (hours)'] = pd.to_numeric(df['Sleep Duration (hours)'], errors='coerce') 
```

```python
df['Sleep Duration (hours)'].describe()
```

```python
import numpy as np
```

```python
df.loc[df['Sleep Duration (hours)'] < 1, 'Sleep Duration (hours)'] = np.nan
```

```python
df['Sleep Duration (hours)'] = df['Sleep Duration (hours)'].fillna(df['Sleep Duration (hours)'].mean())
```

```python
df.info()
```

```python
df['Sleep Duration (hours)'].dtypes
```

```python
df['Activity Level'] = df['Activity Level'].str.replace('Highly_Active', 'Highly Active')
df['Activity Level'] = df['Activity Level'].str.replace('Actve', 'Active')
df['Activity Level'] = df['Activity Level'].str.replace('Seddentary', 'Sedentary')
```

```python
df['Activity Level'].unique()
```

```python
df['Activity Level'] = df['Activity Level'].fillna(df['Activity Level'].mode()[0])
```

```python
df.info()
```

```python
df[df['Stress Level'].isnull() == True]
```

```python
df['Stress Level'] = pd.to_numeric(df['Stress Level'], errors='coerce')
```

```python
df['Stress Level'].unique()
```

```python
df['Stress Level'].describe()
```

```python
df['Stress Level'].mode()
```

```python
df['Stress Level'] = df['Stress Level'].fillna(2.0)
```

```python
df.info()
```
