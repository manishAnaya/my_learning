## Changing Sleep Duration, user_id in numeric.
## Changing other numerical columns to decimal upto 2 values.

```python
df['Sleep Duration (hours)'] = pd.to_numeric(df['Sleep Duration (hours)'], errors='coerce')
```

```python
df['Heart Rate (BPM)'].plot(kind='box')
```

```python
df.dropna(subset=['Heart Rate (BPM)'], inplace=True)
```

```python
df.isna().sum()
```

```python
df = df[['Heart Rate (BPM)','Blood Oxygen Level (%)','Step Count','Sleep Duration (hours)', 'Activity Level', 'Stress Level']]
```

```python
numerical_col = ['Heart Rate (BPM)','Blood Oxygen Level (%)','Step Count','Sleep Duration (hours)']
```

```python
for col in numerical_col:
    df[col] = df[col].astype('float32')
```

```python
for col in numerical_col:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.info()
```

```python
df['Activity Level'] = df['Activity Level'].astype('category')
```

```python
df['Stress Level'] = df['Stress Level'].fillna(5, inplace=True)
```

```python
df['Stress Level'] = pd.to_numeric(df['Stress Level'], errors='coerce')
```

```python
df['Stress Level'] = df['Stress Level'].astype('int32')
```

```python
df.info()
```

```python
df.head()
```

```python
df['Activity Level'].unique()
```

```python
df['Activity Level'] = df['Activity Level'].replace('Actve', 'Active')
```

```python
df['Activity Level'] = df['Activity Level'].replace('Seddentary', 'Sedentary')
```

```python
df['Activity Level'] = df['Activity Level'].replace(
    'Highly_Active',
    'Highly Active'
)
```

```python
df['Activity Level'] = df['Activity Level'].str.strip()
```

```python
df['Activity Level'] = df['Activity Level'].fillna(
    df['Activity Level'].mode()[0]
)
```

```python
df.info()
```

```python
df.head(2)
```

```python
numerical_col = ['Heart Rate (BPM)','Blood Oxygen Level (%)','Step Count','Sleep Duration (hours)']
```

```python
for col in numerical_col:
    df[col] = df[col].round(2)
```

```python
df['Step Count'] = df['Step Count'].round(0).astype('int32')
```

```python
df['Heart Rate (BPM)'] = df['Heart Rate (BPM)'].round(0).astype('int32')
```

```python
df['Blood Oxygen Level (%)'] = (
    df['Blood Oxygen Level (%)']
    .astype('float64')
    .round(2)
)
```
