```python
def changeEmbark(value):
    if value == 'Q':
        return 'Quattro'
    elif value == 'S':
        return 'Supremo'
    elif value == 'C':
        return 'Classsico'
    else:
        return value
```

```python
pd.read_csv('titanic.csv', converters={'Embarked': changeEmbark})
```
