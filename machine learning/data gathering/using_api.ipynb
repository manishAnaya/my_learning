```python
import pandas as pd
```

```python
main_df = pd.DataFrame()
```

```python
for i in range(1, 501):

    url = f'https://api.themoviedb.org/3/trending/movie/day?language=en-US?page={i}'.format(i)
    response = requests.get(url=url, headers=headers)
    data = response.json()['results']

    df = pd.DataFrame(data)
    df = df[['id', 'title', 'poster_path', 'overview', 'adult', 'original_language', 'popularity', 'release_date', 'vote_average', 'vote_count']]

    main_df = pd.concat([main_df, df], ignore_index=False)
```
