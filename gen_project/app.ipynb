```python
from langchain_ollama import OllamaEmbeddings
```

```python
embeddings = OllamaEmbeddings(model="nomic-embed-text")
```

```python
query = "How can I search for text with a similar meaning?"
```

```python
query_vec = embeddings.embed_query(query)
```

```python
len(embeddings.embed_query(query))
```

```python
documents = [
    "Ollama lets you run large language models locally on your computer.",
    "LangChain connects language models with tools, prompts, and document retrieval.",
    "OllamaEmbeddings converts text into vectors for semantic search.",
    "Cosine similarity compares vectors to find text with similar meaning.",
]
```

```python
vectors = embeddings.embed_documents(documents)
```

```python
len(vectors[0])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
scores = cosine_similarity([query_vec], vectors)
```

```python
sorted(list(scores[0]))[-1]
```

```python
idx = scores.argmax()
```

```python
documents[idx]
```

```python
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
docs = [
    Document(
        page_content="Raghav is a Bollywood actor known for his performances in Hindi-language films.",
        metadata={"actor": "Raghuraj Pratap Vaish", "category": "career"}
    ),
    Document(
        page_content="Raghav began his acting career in Bollywood and has appeared in several feature films.",
        metadata={"actor": "Raghuraj Pratap Vaish", "category": "career"}
    ),
    Document(
        page_content="Raghav is recognized for taking on a range of roles, including dramatic and comedic characters.",
        metadata={"actor": "Raghuraj Pratap Vaish", "category": "acting style"}
    ),
    Document(
        page_content="Raghav has worked with directors and co-stars across the Hindi film industry.",
        metadata={"actor": "Raghuraj Pratap Vaish", "category": "collaborations"}
    ),
]
```

```python
docs
```

```python
vector_store = Chroma(embedding_function=embeddings, persist_directory='chroma_db', collection_name='sample_data')
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['documents', 'embeddings', 'metadatas'])
```

```python
vector_store.similarity_search(query='What kind of roles is Raghav known for', k=1)
```

```python
vector_store.similarity_search_with_score(query='What kind of roles is Raghav known for')
```

```python
new_doc = Document(
    page_content='Raghav is recognized for taking on a range of roles, including dramatic and action',
    metadata={"actor": "Raghuraj Pratap Vaish", "category": "acting style"}
)
```

```python
vector_store.update_document(document_id='0053ba94-a207-4c3f-b89e-ef4f605830ec', document=new_doc)
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
embeddings = OllamaEmbeddings(model='nomic-embed-text')
```

```python
db = FAISS.from_documents(docs, embeddings)
```

```python
db
```

```python
db.similarity_search(query='What kind of roles is Raghav known for', k=1)
```

```python
db.similarity_search_with_score(query='What kind of roles is Raghav known for')
```
