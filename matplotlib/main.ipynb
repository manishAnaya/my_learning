```python
monthly_sales = sales_df.groupby(
    [
        sales_df['Date'].dt.year.rename('Year'),
        sales_df['Date'].dt.month.rename('Month')
    ]
)['Amount'].sum().reset_index()

monthly_sales
```

```python
monthly_sales['Month_Name'] = pd.to_datetime(monthly_sales['Month'], format='%m').dt.strftime('%b')
```

```python
for year in monthly_sales['Year'].unique():
    data = monthly_sales[monthly_sales["Year"] == year]
    plt.plot(
        data['Month_Name'],
        data['Amount'],
        marker = 'o'
    )
    
plt.title("Monthly Sales by Year")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.legend()
plt.grid()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
d_frame = pd.read_csv('sales_data.csv')
```

```python
d_frame.info()
```

```python
sales_data = d_frame.groupby(['Sales_Rep', 'Product_Category'])['Sales_Amount'].sum().reset_index()
sales_data
```

```python
categories = sales_data['Product_Category'].unique()

for category in categories:
    print(sales_data[sales_data['Product_Category'] == category])
```

```python
plt.figure(figsize=(10, 8))

sales_persons = sales_data['Sales_Rep'].unique()
categories = sales_data['Product_Category'].unique()
bottom = np.zeros(len(sales_persons))
x = np.arange(len(sales_persons))
for category in categories:
    data = sales_data[sales_data['Product_Category'] == category]
    plt.bar(
        x,
        data['Sales_Amount'],
        bottom = bottom,
        label = category
    )
    bottom += data['Sales_Amount'].values
plt.title('Sales by Sales Representative and Product Category')
plt.xlabel('Sales Representative')
plt.ylabel('Sales Amount')
plt.legend()
plt.show()
```
