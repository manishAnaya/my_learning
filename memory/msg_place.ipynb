```python
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b')

parser = StrOutputParser()

prompt = ChatPromptTemplate.from_messages([
	('system', 'You are helpful AI Assistant with Specialisation in Mathematics, You will be having chat past history So always use to continue the chat naturally'),
    MessagesPlaceholder(variable_name='history'),
	('human', '{input}')
])
```

```python
store = {}

def get_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]
```

```python
chain = prompt | llm | parser
```

```python
chat_with_history = RunnableWithMessageHistory(
    chain,
    get_history,
    input_messages_key='input',
    history_messages_key='history'
)
```

```python
res1 = chat_with_history.invoke(
    {
        "input": "Hello I am Manish",
    },
    config={"configurable": {'session_id': 'manish123'}}
)
res1
```

```python
res2 = chat_with_history.invoke(
    {
        "input": "ok if a = 1, b= 2 and so on what will be names total value.. solve step by step",
    },
    config={"configurable": {'session_id': 'manish123'}}
)
res2
```
