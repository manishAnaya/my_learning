```python
from langchain_core.messages import SystemMessage, trim_messages, AIMessage, HumanMessage
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from operator import itemgetter
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
model = ChatGroq(model='open/gpt-oss-120b')
```

```python
trimmer = trim_messages(max_tokens=50, token_counter=model, include_system=True, allow_partial=False, start_on='human')
```

```python
messages = [
    HumanMessage(content="Hi, I am learning Python. What is a list?"),
    AIMessage(content="A list is an ordered, changeable collection of items, written with square brackets."),
    HumanMessage(content="How do I add an item to a list?"),
    AIMessage(content="Use the append() method, for example: fruits.append('mango')."),
    HumanMessage(content="How can I get the first item?"),
    AIMessage(content="Use index 0: fruits[0]. Python uses zero-based indexing."),
]
```

```python
trimmer.invoke(messages)
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b')

parser = StrOutputParser()

prompt = ChatPromptTemplate.from_messages([
	('system', 'You are helpful AI Assistant with Specialisation in Mathematics, You will be having chat past history So always use to continue the chat naturally'),
    MessagesPlaceholder(variable_name='history'),
	('human', '{input}')
])
```

```python
chain = (RunnablePassthrough.assign(history=itemgetter('history') | trimmer)  | prompt | llm | parser)
```

```python
res = chain.invoke(
    {
        'history': messages,
        'input': 'what I am learning?'
    }
)
print(res)
```
